```python
import pandas as pd

# Load the CSV file into a DataFrame
df = pd.read_csv("WS.csv", header=None, names=["FlowSize", "CDF"])

# Get the lists of flow sizes and CDF values
flow_sizes = df["FlowSize"].tolist()
cdf_values = df["CDF"].tolist()

# Print the results
print("Flow Sizes:", flow_sizes)
print("CDF Values:", cdf_values)
```

```python
total_vol = 0
large_vol = 0
for idx in range(1, len(flow_sizes)):
    total_vol += flow_sizes[idx]*(cdf_values[idx]-cdf_values[idx-1])
    if flow_sizes[idx] >= 60000000:
        large_vol += flow_sizes[idx]*(cdf_values[idx]-cdf_values[idx-1])

print(total_vol)
print(large_vol)
print(large_vol/total_vol*100)
```
